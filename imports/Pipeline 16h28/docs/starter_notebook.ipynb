## Validation helper

Use it for every decision. The leaderboard is macro-F1.

```python
def evaluate(model, X, y, seed=0):
    X_tr, X_val, y_tr, y_val = train_test_split(X, y, test_size=0.2, random_state=seed, stratify=y)
    model.fit(X_tr, y_tr)
    return f1_score(y_val, model.predict(X_val), average="macro")
```
